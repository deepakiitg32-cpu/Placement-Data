```python
import math

def manning_discharge(B, H, n, S):
    A = B * H
    P = B + 2 * H
    R = A / P
    Q = (1 / n) * A * (R ** (2 / 3)) * (S ** 0.5)
    return Q, A

def check_velocity(Q, A):
    V = Q / A
    return V

def culvert_design(Q_required, B, H, n, S, canal_width, spacing=0.5, max_velocity=3.0, min_freeboard=0.3):
    Q_one, A = manning_discharge(B, H, n, S)
    N = math.ceil(Q_required / Q_one)

    total_width_required = N * B + (N - 1) * spacing

    if total_width_required <= canal_width:
        placement = "Place all culverts at the same location."
    else:
        placement = "Distribute culverts at different locations along the canal."

    # IRC Checks
    velocity = check_velocity(Q_one, A)
    velocity_ok = velocity <= max_velocity

    freeboard_ok = (H - (Q_required / (B * N))) >= min_freeboard

    design_summary = {
        "Required Discharge (m³/s)": Q_required,
        "Discharge per Culvert (m³/s)": round(Q_one, 2),
        "Number of Culverts Needed": N,
        "Canal Width (m)": canal_width,
        "Total Width Needed (m)": round(total_width_required, 2),
        "Velocity (m/s)": round(velocity, 2),
        "Velocity OK (<= 3.0 m/s)": velocity_ok,
        "Freeboard OK (>= 0.3 m)": freeboard_ok,
        "Placement Strategy": placement
    }

    return design_summary

# ---- CLI INPUT SECTION ----

def main():
    print("\n BOX CULVERT DESIGN CALCULATOR (Based on IRC Guidelines)\n")

    Q_required = float(input("Enter design discharge (m³/s): "))
    B = float(input("Enter culvert width B (m): "))
    H = float(input("Enter culvert height H (m): "))
    n = float(input("Enter Manning's roughness coefficient (n): "))
    S = float(input("Enter bed slope (S): "))
    canal_width = float(input("Enter total canal width (m): "))

    result = culvert_design(Q_required, B, H, n, S, canal_width)
    print(f"\n Number of Culverts Required: {result['Number of Culverts Needed']}")


    print("\n--- DESIGN SUMMARY ---")
    for key, value in result.items():
        print(f"{key}: {value}")

if __name__ == "__main__":
    main()

```
